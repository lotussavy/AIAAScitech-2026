# Airport taxi and UAM costs

Independent interface to the shared model. Starts from bundled hourly summaries, with durations already in hours. Change `AIRPORT` to EWR, LGA, or JFK. The output uses the plot-compatible schema. See `docs/REPRODUCIBILITY.md` for the historical EWR distance issue.


```python
from pathlib import Path
import sys
import json
import pandas as pd

ROOT = Path.cwd().resolve()
if not (ROOT / "uam_demand").is_dir():
    ROOT = ROOT.parent
if not (ROOT / "uam_demand").is_dir():
    raise RuntimeError("Open this notebook from the repository or notebooks directory")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
CONFIG = json.loads((ROOT / "config.json").read_text())
```

```python
from uam_demand.model import compute_costs
AIRPORT = "EWR"
hourly = pd.read_csv(ROOT / f"data/hourly/{AIRPORT}-hourly.csv")
distances = pd.read_csv(ROOT / "data/taxi_zone_airport_distances.csv")
final_df = compute_costs(hourly, distances, AIRPORT, CONFIG)
output = ROOT / "outputs" / "notebook_costs"
output.mkdir(parents=True, exist_ok=True)
final_df.to_csv(output / f"Final_{AIRPORT}_UAM_GCT.csv", index=False)
final_df.head()
```
